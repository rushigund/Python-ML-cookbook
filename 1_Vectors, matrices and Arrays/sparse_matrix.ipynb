```python
import numpy as np
import scipy.sparse as sparse   
```

```python
sparse_matrix = np.array([[0,0,3],
                        [0,5,0],
                        [7,0,0]])
```

```python
sparse_matrix_1 = sparse.csr_matrix(sparse_matrix)
print(sparse_matrix_1)
```

```python
print(sparse_matrix_1[0,2])
```

```python
print(sparse_matrix_1.ndim)
print(sparse_matrix_1.shape)
print(sparse_matrix_1.size)
print(np.max(sparse_matrix_1,axis=0))
print(np.mean(sparse_matrix_1,axis=1))
print(np.std(sparse_matrix,axis=1))
print(np.var(sparse_matrix))
```

reshape method

```python
print(sparse_matrix.reshape(1,-1))
print(sparse_matrix.flatten())
```

Transpose of matrix

```python
print(sparse_matrix.T)
```

 <font size="6">Rank of matrix</font>

```python
print(np.linalg.matrix_rank(sparse_matrix))
print(np.linalg.det(sparse_matrix))
```

*Diagonal of matrix*

```python
sparse_matrix.diagonal()
sparse_matrix.diagonal(offset=1) #diagonal above main diagonal
sparse_matrix.diagonal(offset=-1) #diagonal below main diagonal
```

<h3>Trace (sum of diagnal elements) of matrix</h3>

```python
eigenvalues, eigenvectors = np.linalg.eig(sparse_matrix)
print("Eigenvalues:", eigenvalues)  
print("Eigenvectors:\n", eigenvectors)
sparse_matrix @ sparse_matrix
```

<font size = "5">Inverse </font>

```python
print(np.linalg.inv(sparse_matrix))
```
