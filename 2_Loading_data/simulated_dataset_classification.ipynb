```python
from sklearn.datasets import make_classification
```

```python
features, targets = make_classification(n_samples=100,
                                        n_features = 3,
                                        n_informative= 3,
                                        n_redundant=0,
                                        n_classes=2,
                                        weights=[0.7, 0.3],
                                        random_state=1)
```

```python
print(features)
print(targets)
```
