```python
from sklearn.datasets import make_blobs
```

```python
features, targets = make_blobs(n_samples=100,
                              n_features=2,         
                              centers=3,
                              cluster_std=1.0,
                              random_state=1)
```

```python
features
```

```python
targets
```

```python
import matplotlib.pyplot as plt
plt.scatter(features[:,0], features[:, 1], c=targets)  
plt.show()
```
